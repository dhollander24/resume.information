Hi! This notebook contains my Exploratory Data Analysis (EDA) for the Ames Housing dataset. The goal of this project is to figure out which features of a house are the best predictors of its final sale price.

## Step 1: Loading the Data
First, we need to import our libraries and load the dataset into a Pandas DataFrame. I downloaded the `train.csv` file from Kaggle.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visualization style
sns.set_theme(style="whitegrid")

# Load the dataset
df = pd.read_csv('train.csv')

# What variables are we looking at?
key_variables = ['SalePrice', 'GrLivArea', 'BedroomAbvGr', 'FullBath', 'YearBuilt']
df_clean = df[key_variables].dropna()

print(f"We are analyzing {len(df_clean)} homes.")
df_clean.head()
```

## Step 2: Basic Statistics
Let's look at some basic summary statistics for our key variables to get a feel for the data.

```python
# Display summary stats nicely
df_clean.describe().round(2)
```

## Step 3: Visualizing the Relationships
Are larger homes really more expensive? Let's plot Sale Price against Above Ground Living Area (Square Footage) to see if my first hypothesis is correct.

```python
plt.figure(figsize=(10, 6))
sns.regplot(data=df_clean, x='GrLivArea', y='SalePrice', 
            scatter_kws={'alpha':0.5, 'color':'steelblue'},
            line_kws={'color':'red', 'linewidth':2})

plt.title('Hypothesis 1: Does Square Footage Predict Price?', fontsize=16)
plt.xlabel('Above Ground Living Area (Sq Ft)', fontsize=12)
plt.ylabel('Sale Price ($)', fontsize=12)

# Format y-axis to show dollar amounts
current_values = plt.gca().get_yticks()
plt.gca().set_yticklabels(['${:,.0f}'.format(x) for x in current_values])

plt.show()
```

The trend line shows a clear positive relationship between square footage and price. My first hypothesis is supported by the data.

## Step 4: Correlation Matrix
Scatter plots are great, but let's get some hard numbers. We can use a correlation matrix to see exactly how strongly each of our variables correlates with `SalePrice`.

```python
plt.figure(figsize=(8, 6))
correlation_matrix = df_clean.corr()

# Create a nice visually appealing heatmap
sns.heatmap(correlation_matrix, 
            annot=True, 
            cmap='coolwarm', 
            center=0, 
            fmt='.2f', 
            linewidths=1)

plt.title('Correlation between Housing Features', fontsize=14)
plt.show()
```

## Preliminary Conclusions
Based on this initial exploration:

1. **Square Footage** is our strongest predictor (correlation = 0.71). Larger homes definitely sell for more.
2. **Bathrooms** (0.56) and **Age** (0.52) are also strong predictors. People pay a premium for newer homes with more bathrooms.
3. **Bedrooms** (0.17) showed a weaker direct correlation. This means we will need a multiple regression model for our final report to figure out how bedrooms affect price when controlling for total square footage!
